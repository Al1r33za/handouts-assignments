## CBC IND-CPA test \w predictable Nonce (=counter, timer)

Because Nonce is predictable Adversary can query $m_0$:
$$
m_0 = x \oplus Nonce \to c_0 = E((x \oplus Nonce)\oplus Nonce) = E(x)
$$
Then Adversary tries
$$
m_0 = x \to c = E(x)
$$
$$
m_1 = y \ne x \to c = E(y)
$$
So Attacker can Distinguish $m_0, m_1 $ by knowing $E(x)$. 

## Modes of Operation CTR \& CBC

### Last block of plaintext is random:

If mode of operation is **CTR** Attacker can change last random block and Alice wouldn't notice it, and also if the mode of operation is **CBC** by decryption only last block will be changed so *as long as attacker not touching blocks other than that random plaintext* alice wont notice the changes.

### Chaning first block of message

Attacker can replace $M_0$ with $A_1$ in **CTR**, since in **CTR** blocks are separated from each other this wont affect other blocks. in **CBC** case, changing first block would change $c_1$ but if we decrypt with unchanged or changed ciphertext it wouldn't make any difference, changed plaintext will be $A_1$ and other blocks would be untouched.
