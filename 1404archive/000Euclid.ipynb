```python

# Euclid Algorith 
# gcd(r0, r1) = gcd(r1, r0 mod r1) 
def gcdE(r0, r1):
    ''' Euclid Algorithm for gcd
        gcd(r0, r1) = gcd(r1, r0 mod r1)'''
    if(r0 < r1):
        r0, r1 = r1, r0
    if(r1 == 0):
        return r0
    else:
        return gcdE(r1, r0 % r1)

# extended Euclid Algorithm
# used to compute INVERSE MODULATION: 1 = t.r1 mod r0
def EEA(r0, r1, s0=1, s1=0, t0=0, t1=1):
    '''Extended Euclid Algorithm 
       r[i] = s[i]r[0] + t[i]r[1]
       r[i] = r[i-2] mod r[i-1]
       s[i] = s[i-2] + q[i-1]s[i-1]
       t[i] = t[i-2] + q[i-1]s[i-1]
       q[i-1] = (r[i-2]-r[i])/r[i-1]'''
    
    if(r0 < r1):
        r0, r1 = r1, r0
    if(r1 == 0):
        return r0, s0, t0
    q = r0 // r1
    s2 = s0 - q*s1
    t2 = t0 - q*t1
    return EEA(r1, r0 % r1, s1, s2, t1, t2)

def invmod(a, m):
    '''computes : a^(-1) mod m  a < m'''
    return EEA(a, m)[2] % m

```
