## CBC-MAC with Variable-Length Messages

We consider two constructions for handling variable-length messages with CBC-MAC.

### (a) Length Prepending

The construction is:

$$
\text{CBC-MAC}(\text{length} \parallel M_1 \parallel M_2 \parallel \cdots \parallel M_n)
$$

**Answer:**  
This construction **is secure** against length-extension attacks.

**Reason:**  
Standard CBC-MAC is only secure for *fixed-length* messages.  
By prepending the length of the message, all messages effectively become unique encodings with respect to their length. This prevents an attacker from reusing a valid MAC of one message to forge the MAC of another message with a different length.

Hence, length-prepending converts variable-length messages into a form that satisfies the fixed-length requirement of CBC-MAC.

### (b) Length Appending

The alternative construction is:

$$
\text{CBC-MAC}(M_1 \parallel M_2 \parallel \cdots \parallel M_n \parallel \text{length})
$$

**Reason:**  
Appending the length at the end does **not** prevent length-extension attacks.  
An attacker can exploit the iterative structure of CBC-MAC to construct valid MACs for longer messages by extending previously authenticated messages and adjusting the final blocks.

Therefore, **length-appending does not fix the variable-length insecurity of CBC-MAC**.

## Iterative Hash Function Based on AES

The hash construction is defined as:

$$
H_0 = 0,\quad
H_i = \text{AES}_{M_i}(H_{i-1}) \quad \text{for } i=1,\ldots,\ell,\quad
T = H_\ell
$$
### (a) Attack Description

**Attack:**  
This construction is **not collision resistant**.

**Explanation:**  
AES is a *permutation* for a fixed key.  
Since the **message block is used as the AES key**, an attacker can choose different message blocks \( M_i \) that map the same input chaining value \( H_{i-1} \) to the same output \( H_i \).

This allows the attacker to construct **collisions efficiently**, violating a fundamental requirement of cryptographic hash functions.

### (b) Replacing AES with a Compression Function

Now the construction is modified to:

$$
H_i = \text{AES}_{H_{i-1}}(M_i)
$$

Security is **improved**, but it does not reach the robustness of standard hash constructions such as SHA-2 or SHA-3.

## MAC Construction Using a Hash Function

The MAC is defined as:

$$
\text{MAC}_K(M) = H(K \parallel M)
$$

### Attack Description (Length Extension Attack)

**Attack:**  
If the hash function \( H \) is based on a Merkle–Damgård construction (e.g., SHA-256), this MAC is **insecure**.

**Explanation:**  
Given:
$$
H(K \parallel M)
$$

An attacker who knows the MAC value can compute:
$$
H(K \parallel M \parallel M')
$$

**without knowing the key** $ K $, using the length-extension property of Merkle–Damgård hashes.

Thus, the attacker can forge valid MACs for extended messages.
